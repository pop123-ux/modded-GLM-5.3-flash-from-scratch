# Local experiment quickstart

Run in VS Code or Jupyter with the cloned repository (or extracted project) available and its `.venv` selected as your kernel. Follow [IDE_GUIDE.md](../IDE_GUIDE.md) for setup. Theory notebooks are primarily reading material; this notebook is a runnable CPU smoke check, not a convergence benchmark.

```python
from pathlib import Path
import sys, subprocess, json, datetime, os
ROOT = next((p for p in [Path.cwd(), *Path.cwd().parents] if (p/'pyproject.toml').exists() and (p/'glm53_flash').is_dir()), None)
assert ROOT is not None, 'Open Jupyter/VS Code inside the repository folder.'
os.chdir(ROOT)
print('Repository:', ROOT)
print('Kernel:', sys.executable)
```

```python
OUTPUT = ROOT/'runs'/('local-notebook-' + datetime.datetime.now(datetime.timezone.utc).strftime('%Y%m%d-%H%M%S-%f'))
subprocess.run([sys.executable,'scripts/train_pretrain.py','--config','configs/tiny.json','--output',str(OUTPUT),'--steps','3','--batch-size','2','--eval-per-family','1','--device','cpu'],check=True,cwd=ROOT)
```

```python
receipt = json.loads((OUTPUT/'training-receipt.json').read_text(encoding='utf-8'))
print('Status:', receipt['status'])
print('Final dev loss:', receipt['validation_curve'][-1]['val_loss'])
print('Receipt:', OUTPUT/'training-receipt.json')
```
